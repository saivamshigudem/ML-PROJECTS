```python
from sklearn.preprocessing import StandardScaler
```

```python
print("\n")

print("="*70)
print("Missing Values Before Treatment")
print("="*70)

display(df.isnull().sum())
```

```python
for column in df.columns:

    if df[column].dtype == "object":

        df[column].fillna(

            df[column].mode()[0],

            inplace=True

        )

    else:

        df[column].fillna(

            df[column].median(),

            inplace=True

        )

print("\n")

print("="*70)
print("Missing Values After Treatment")
print("="*70)

display(df.isnull().sum())
```

```python
X = df.drop(

    "label",

    axis=1

)

y = df["label"]

print("\n")

print("="*70)
print("Feature Matrix and Target Variable")
print("="*70)

print("Feature Shape :", X.shape)

print("Target Shape  :", y.shape)
```

```python
print("\n")

print("="*70)
print("Feature Matrix")
print("="*70)

display(X.head())
```

```python
print("\n")

print("="*70)
print("Target Variable")
print("="*70)

display(y.head())
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(

    X

)

print("\n")

print("="*70)
print("Feature Scaling Completed")
print("="*70)

print("Scaled Dataset Shape :", X_scaled.shape)
```

```python
X_scaled_df = pd.DataFrame(

    X_scaled,

    columns=X.columns

)

print("\n")

print("="*70)
print("Scaled Feature Matrix")
print("="*70)

display(X_scaled_df.head())
```

```python
print("\n")

print("="*70)
print("Mean of Scaled Features")
print("="*70)

display(

    X_scaled_df.mean().round(2)

)
```

```python
print("\n")

print("="*70)
print("Standard Deviation of Scaled Features")
print("="*70)

display(

    X_scaled_df.std().round(2)

)
```

```python
sample_size = 5000

X_sample = X_scaled_df.sample(

    n=sample_size,

    random_state=42

)

y_sample = y.loc[

    X_sample.index

]

print("\n")

print("="*70)
print("Sample Dataset Created")
print("="*70)

print("Sample Feature Shape :", X_sample.shape)

print("Sample Target Shape  :", y_sample.shape)
```

```python
X_sample.reset_index(

    drop=True,

    inplace=True

)

y_sample.reset_index(

    drop=True,

    inplace=True

)
```

```python
print("\n")

print("="*70)
print("Sample Class Distribution")
print("="*70)

display(

    y_sample.value_counts().sort_index()

)

plt.figure(figsize=(8,5))

sns.countplot(

    x=y_sample

)

plt.title("Digit Distribution in Sample Dataset")

plt.xlabel("Digit")

plt.ylabel("Count")

plt.show()
```

```python
import time

from sklearn.decomposition import PCA

from sklearn.manifold import TSNE

import umap.umap_ as umap
```

```python
start_time = time.time()

pca = PCA(

    n_components=2,

    random_state=42

)

X_pca = pca.fit_transform(

    X_sample

)

pca_time = time.time() - start_time

print("\n")

print("="*70)

print("PCA Completed Successfully")

print("="*70)

print("Execution Time :", round(pca_time,2),"Seconds")
```

```python
print("\n")

print("="*70)

print("PCA Explained Variance")

print("="*70)

for i,value in enumerate(

    pca.explained_variance_ratio_

):

    print(

        f"PC{i+1} : {value:.4f}"

    )

print(

    "\nTotal Variance :",

    round(

        pca.explained_variance_ratio_.sum(),

        4

    )

)
```

```python
pca_df = pd.DataFrame(

    X_pca,

    columns=["PC1","PC2"]

)

pca_df["Label"] = y_sample

display(

    pca_df.head()

)
```

```python
print("\n")

print("="*70)

print("PCA Explained Variance")

print("="*70)

for i,value in enumerate(

    pca.explained_variance_ratio_

):

    print(

        f"PC{i+1} : {value:.4f}"

    )

print(

    "\nTotal Variance :",

    round(

        pca.explained_variance_ratio_.sum(),

        4
          )

)
```

```python
plt.figure(figsize=(10,7))

sns.scatterplot(

    data=pca_df,

    x="PC1",

    y="PC2",

    hue="Label",

    palette="tab10",

    s=25,

    alpha=0.8

)

plt.title("PCA Visualization")

plt.show()
```

```python
start_time = time.time()

tsne = TSNE(

    n_components=2,

    perplexity=30,

    learning_rate="auto",

    init="pca",

    random_state=42

)

X_tsne = tsne.fit_transform(

    X_sample

)

tsne_time = time.time() - start_time

print("\n")

print("="*70)

print("t-SNE Completed Successfully")

print("="*70)

print("Execution Time :", round(tsne_time,2),"Seconds")
```

```python
tsne_df = pd.DataFrame(

    X_tsne,

    columns=["Dim1","Dim2"]

)

tsne_df["Label"] = y_sample

display(

    tsne_df.head()

)
```

```python
plt.figure(figsize=(10,7))

sns.scatterplot(

    data=tsne_df,

    x="Dim1",

    y="Dim2",

    hue="Label",

    palette="tab10",

    s=25,

    alpha=0.8

)

plt.title("t-SNE Visualization")

plt.show()
```

```python
start_time = time.time()

umap_model = umap.UMAP(

    n_components=2,

    n_neighbors=15,

    min_dist=0.1,

    random_state=42

)

X_umap = umap_model.fit_transform(

    X_sample

)

umap_time = time.time() - start_time

print("\n")

print("="*70)

print("UMAP Completed Successfully")

print("="*70)

print("Execution Time :", round(umap_time,2),"Seconds")
```

```python
umap_df = pd.DataFrame(

    X_umap,

    columns=["UMAP1","UMAP2"]

)

umap_df["Label"] = y_sample

display(

    umap_df.head()

)
```

```python
plt.figure(figsize=(10,7))

sns.scatterplot(

    data=umap_df,

    x="UMAP1",

    y="UMAP2",

    hue="Label",

    palette="tab10",

    s=25,

    alpha=0.8

)

plt.title("UMAP Visualization")

plt.show()
```

```python
comparison = pd.DataFrame({

    "Algorithm":[

        "PCA",

        "t-SNE",

        "UMAP"

    ],

    "Execution Time (Seconds)":[

        round(pca_time,2),

        round(tsne_time,2),

        round(umap_time,2)

    ]

})

print("\n")

print("="*70)

print("Execution Time Comparison")

print("="*70)

display(comparison)
```

```python
plt.figure(figsize=(8,5))

sns.barplot(

    data=comparison,

    x="Algorithm",

    y="Execution Time (Seconds)"

)

plt.title("Execution Time Comparison")

plt.show()
```
