```python
from sklearn.preprocessing import StandardScaler
```

```python
print("\n")

print("="*70)
print("Missing Values Before Treatment")
print("="*70)

display(df.isnull().sum())
```

```python
for column in df.columns:

    if df[column].dtype == "object":

        df[column].fillna(

            df[column].mode()[0],

            inplace=True

        )

    else:

        df[column].fillna(

            df[column].median(),

            inplace=True

        )

print("\n")

print("="*70)
print("Missing Values After Treatment")
print("="*70)

display(df.isnull().sum())
```

```python
X = df.drop(

    "Class",

    axis=1

)

y = df["Class"]

print("\n")

print("="*70)
print("Feature Matrix and Target Variable")
print("="*70)

print("Feature Shape :", X.shape)

print("Target Shape  :", y.shape)
```

```python
print("\n")

print("="*70)
print("Feature Matrix")
print("="*70)

display(X.head())
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(

    X

)

print("\n")

print("="*70)
print("Feature Scaling Completed")
print("="*70)

print("Scaled Data Shape :", X_scaled.shape)
```

```python
X_scaled_df = pd.DataFrame(

    X_scaled,

    columns=X.columns

)

print("\n")

print("="*70)
print("Scaled Feature Matrix")
print("="*70)

display(X_scaled_df.head())
```

```python
print("\n")

print("="*70)
print("Dataset Summary")
print("="*70)

print("Total Records          :", X.shape[0])

print("Total Features         :", X.shape[1])

print("Fraud Transactions     :", y.sum())

print("Normal Transactions    :", len(y)-y.sum())

print("Fraud Percentage (%)   :", round(y.mean()*100,4))

print("Scaled Dataset Shape   :", X_scaled.shape)

print("\nDataset is Ready for Isolation Forest Training.")

print("="*70)
```

```python
from sklearn.ensemble import IsolationForest

from sklearn.metrics import (

    confusion_matrix,

    classification_report,

    accuracy_score,

    precision_score,

    recall_score,

    f1_score,

    roc_auc_score

)
```

```python
model = IsolationForest(

    n_estimators=100,

    contamination=0.0017,

    random_state=42,

    n_jobs=-1

)

print("\n")

print("="*70)

print("Isolation Forest Model Created Successfully")

print("="*70)
```

```python
model.fit(

    X_scaled

)

print("\n")

print("="*70)

print("Model Training Completed Successfully")

print("="*70)
```

```python
predictions = model.predict(

    X_scaled

)

print("\n")

print("="*70)

print("Predictions Generated Successfully")

print("="*70)
```

```python
predicted_class = np.where(

    predictions == -1,

    1,

    0

)
```

```python
anomaly_score = -model.decision_function(

    X_scaled

)

df["Anomaly_Score"] = anomaly_score

df["Predicted_Class"] = predicted_class
```

```python
print("\n")

print("="*70)

print("Prediction Summary")

print("="*70)

summary = pd.DataFrame({

    "Count": pd.Series(predicted_class).value_counts()

})

display(summary)
```

```python
fraud_percentage = (

    predicted_class.sum()

    /

    len(predicted_class)

)*100

print("\n")

print("="*70)

print("Detected Fraud Percentage")

print("="*70)

print(

    round(fraud_percentage,4),

    "%"

)
```

```python
cm = confusion_matrix(

    y,

    predicted_class

)

print("\n")

print("="*70)

print("Confusion Matrix")

print("="*70)

display(cm)

plt.figure(figsize=(6,5))

sns.heatmap(

    cm,

    annot=True,

    fmt="d",

    cmap="Blues",

    xticklabels=["Normal","Fraud"],

    yticklabels=["Normal","Fraud"]

)

plt.xlabel("Predicted")

plt.ylabel("Actual")

plt.title("Confusion Matrix")

plt.show()
```

```python
print("\n")

print("="*70)

print("Classification Report")

print("="*70)

print(

    classification_report(

        y,

        predicted_class

    )

)
```

```python
accuracy = accuracy_score(

    y,

    predicted_class

)

precision = precision_score(

    y,

    predicted_class,

    zero_division=0

)

recall = recall_score(

    y,

    predicted_class,

    zero_division=0

)

f1 = f1_score(

    y,

    predicted_class,

    zero_division=0

)

roc_auc = roc_auc_score(

    y,

    predicted_class

)

metrics = pd.DataFrame({

    "Metric":[

        "Accuracy",

        "Precision",

        "Recall",

        "F1 Score",

        "ROC-AUC"

    ],

    "Value":[

        accuracy,

        precision,

        recall,

        f1,

        roc_auc

    ]

})

print("\n")

print("="*70)

print("Evaluation Metrics")

print("="*70)

display(metrics)
```
